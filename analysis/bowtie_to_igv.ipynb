```python
import pandas as pd
import numpy as np
```

```python
path = "/Users/kalebzuckerman/Desktop/paper_pushing/igv_vis/hs_soap.bed"
fwd_out = "/Users/kalebzuckerman/Desktop/paper_pushing/igv_vis/hs_soap_fwd.bed"
rev_out = "/Users/kalebzuckerman/Desktop/paper_pushing/igv_vis/hs_soap_rev.bed"
```

```python
dat = pd.read_table(path, delimiter="\t", names=["name", "strand", "chr", "start", "primer","score", "sumnum"], index_col=False)
dat = dat.assign(end = dat["start"] + dat["primer"].str.len())
dat = dat.drop(["primer", "score", "sumnum"], axis=1)[~dat["chr"].str.contains("_")]
dat.head()
fwd = dat.loc[dat["strand"] == "+", ["chr","start","end", "name"]]
fwd.to_csv(fwd_out, sep="\t", header=None, index=None)
rev = dat.loc[dat["strand"] == "-", ["chr","start","end", "name"]]
rev.to_csv(rev_out, sep="\t", header=None, index=None)
```
